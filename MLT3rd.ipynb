```python
#imputation
titanic_data.drop('Cabin',axis=1,inplace=True)
```

```python
pd.get_dummies(titanic_data['Sex']).astype(int)
```

```python
S=pd.get_dummies(titanic_data['Sex']).astype(int)
S.head(4)
```

```python
#converts of embarked column from one form to another
#Q=Quinstown,S=southam,C=chambor
E=pd.get_dummies(titanic_data['Embarked'],drop_first=True).astype(int).astype(int)
E.head(4)
```

```python
#convertion of pclass column from one form to another
P=pd.get_dummies(titanic_data['Pclass'],drop_first=True).astype(int)
P.head(4)
```

```python
#concating all the new rows in the dataset
titanic_data=pd.concat([titanic_data,S,E,P],axis=1)
```

```python
titanic_data.head(5)
```

```python
titanic_data.drop(['PassengerId','Name','Sex','Ticket','Embarked','Pclass'],axis=1,inplace=True)
```

```python
#our final dataset
titanic_data.head(5)
```

```python
X=titanic_data.drop('Survived',axis=1)#accept survied all the other columns are indep var
y=titanic_data['Survived']
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
X_train.columns = X_train.columns.astype(str)
model.fit(X_train,y_train)
```

```python
X_test.columns = X_test.columns.astype(str)
predictions=model.predict(X_test)
```

```python
print("Predictions:")
print(predictions)
```

```python
import matplotlib.pyplot as plt



#Assuming you already have your trained model and test data
#model.fit(X_train,y_train)
predictions=model.predict(X_test)

#plot predicted clasess vs the frist feature of X_test
plt.plot(X_test.iloc[:,0],predictions,marker='o',linestyle='-',color='blue')
plt.xlabel('Features')
plt.ylabel('Survived')
plt.title('Logistc Regression Predictions')
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(y_test,predictions))
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,predictions)
```

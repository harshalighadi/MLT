**Steps In Linear regression**

```python
#here to implement simple linear regression will perform some steps
#step1 : importing required libraries
#step2 : read the Data
#step3 : model selection and fitting
#step4 : prediction the value using linear equation y=mx+b
#step5 : visualizing the data
```

**step1: Importing libraries**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

**Step2: Read The Data**

```python
data={"hours":[1,2,3,4,5],
      "Marks":[50,60,80,90,100]}
df=pd.DataFrame(data)
print(df)
```

```python
X=df[["hours"]].values
y=df[["Marks"]]
```

**Step3: Model Selction And Fitting**

Model Fitting
Equation for linera regression is y=mx+c what is slope it is y/x means,change in dependent variable when there is unit change in independent variable .

what is intercercept= even in the absence of x there will be some value i.e even if the student is not studing he or she will be getting some marks that is basically intercept according to the example

here we are finding the change in marks when their is unit change in no of hours studied

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)
```

```python
slope=model.coef_[0]
intercept=model.intercept_
```

**Step4: Predicting The value using linear equation**

```python
#"f-string" (formatted string literal) in python
#.2f: This is a format specifier used within the curly braces of an f-string
print(f"slope(m):{slope}")
print(f"Intercept(b):{intercept}")
print(f"Equation of the regression line:y={slope[0]:.2f} x + {intercept[0]:.2f}")
```

```python
df['Predicted Marks']=model.predict(X)
```

```python
#here keep in mind the data which we have provided to the model is also liner
df.head()
```

**Step5: Visualizing**

```python
plt.scatter(df['hours'],df['Predicted Marks'],color='red',label='Actual Marks')
```

**Plotting the equation**

```python
plt.plot(df['hours'],df['Predicted Marks'],color='red',label='Regression Line')
plt.xlabel('Hours studied')
plt.ylabel('Marks')
plt.title('Linear Regression:Hours vs Marks')
plt.legend()
plt.show()
```
